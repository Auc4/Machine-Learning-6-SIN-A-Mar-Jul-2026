```python
# Nulls deleted
df = df.dropna()
```

```python
# Sex verification of registries
df["sex"].unique()
```

```python
# Delete of labeled "." sex 
df = df[df['sex'] != "."]
```

```python
# Verication
df['sex'].unique()
```

```python
# Make numeric categorical values
pd.get_dummies(df)
```

```python
pd.get_dummies(df.drop('species',axis=1),drop_first=True)
```

**Explanation:** This commit drops the column `species` and discard the first coicidente for visualize the exact result

```python
X = pd.get_dummies(df.drop('species',axis=1),drop_first=True)
y = df['species']
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=866)
```

# Decision Tree Classifier
## Default Hyperparameters

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(X_train,y_train)

base_pred = model.predict(X_test)
```

# Evaluation

```python
from sklearn.metrics import confusion_matrix,classification_report,ConfusionMatrixDisplay 
cm = confusion_matrix(y_test,base_pred)
```

```python
plt.style.use('classic')

ConfusionMatrixDisplay.from_estimator(model, X_test, y_test)
```

```python
print(classification_report(y_test,base_pred))
```

Support is a metric that explains how actual rows of that class are in y_test.

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(12,8))
plot_tree(model);
```

```python
plt.figure(figsize=(12,8),dpi=150)
plot_tree(model,filled=True,feature_names=X.columns);
```
